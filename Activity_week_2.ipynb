```python
import pandas as pd
import matplotlib.pyplot as plt

# Since the header is missing a comma ('Location PM2.5 Level'), we skip the header row
# and manually assign correct column names.
correct_columns = ['Date', 'Location', 'PM2.5 Level']
df = pd.read_csv('/content/Air_Quality.csv', skiprows=1, names=correct_columns, encoding='utf-8-sig')

# Clean column names and any string values
df.columns = df.columns.astype(str).str.strip()
df['Location'] = df['Location'].astype(str).str.strip()

# --- DATA PREPARATION & AGGREGATION ---

# Filter data for just the city location
city_center_data = df[df['Location'] == 'city']

# Group data by location to find average pollution levels
avg_pollution = df.groupby('Location')[['PM2.5 Level']].mean()

# --- THE 5 REQUIRED CHARTS ---

# Chart 1: Line Chart (Tracking Air Quality over time at city location)
plt.figure(figsize=(8, 4))
plt.plot(city_center_data['Date'], city_center_data['PM2.5 Level'], marker='o', color='red')
plt.title('Chart 1: PM2.5 Level Trend at City Location')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Bar Chart (Comparing average PM 2.5 by location)
plt.figure(figsize=(8, 4))
avg_pollution['PM2.5 Level'].plot(kind='bar', color='orange')
plt.title('Chart 2: Average PM2.5 Particles by Location')
plt.xlabel('Location')
plt.ylabel('Average PM2.5')
plt.xticks(rotation=0)
plt.show()
```

```python
# Chart 3: Scatter Plot (Distribution of PM2.5 Levels over Dates)
plt.figure(figsize=(8, 4))
plt.scatter(df['Date'], df['PM2.5 Level'], color='purple', s=50)
plt.title('Chart 3: PM2.5 Level Distribution by Date')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset using the correct path and configuration from Cell 1
correct_columns = ['Date', 'Location', 'PM2.5 Level']
df = pd.read_csv('/content/Air_Quality.csv', skiprows=1, names=correct_columns, encoding='utf-8-sig')

# Clean column names
df.columns = df.columns.astype(str).str.strip()

# Chart 4: Histogram (Distribution of all PM2.5 readings)
plt.figure(figsize=(8, 4))
plt.hist(df['PM2.5 Level'], bins=5, color='skyblue', edgecolor='black')
plt.title('Chart 4: Frequency of Different PM2.5 Levels')
plt.xlabel('PM2.5 Level')
plt.ylabel('Frequency')
plt.show()
```

```python
# Chart 5: Horizontal Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5 Level'].max().plot(kind='barh', color='teal')
plt.title('Chart 5: Maximum PM2.5 Recorded per Location')
plt.xlabel('Peak PM2.5 Level')
plt.ylabel('Location')
plt.show()
```
